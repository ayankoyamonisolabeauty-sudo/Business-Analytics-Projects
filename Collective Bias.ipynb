```python
import pandas as pd

# Load the dataset using the file in the same directory
data = pd.read_csv('CB- Cleaned Dataset Train.csv')  # Ensure the file is in the same directory as your notebook

# Check the data
data.head()

```

```python
# Check the data structure and column info
data.info()
```

```python
# Get summary statistics of numerical columns
data.describe()
```

```python
# Check for missing values in the dataset
print(data.isnull().sum())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot distribution of client spending
plt.figure(figsize=(8, 6))
sns.histplot(data['client_spend'], kde=True)
plt.title('Distribution of Client Spend')
plt.show()

# Plot distribution of total engagement
plt.figure(figsize=(8, 6))
sns.histplot(data['total_engagement'], kde=True)
plt.title('Distribution of Total Engagement')
plt.show()
```

```python
# Outlier Detection using IQR for Client Spend
Q1 = data['client_spend'].quantile(0.25)
Q3 = data['client_spend'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
outliers = data[(data['client_spend'] < lower_bound) | (data['client_spend'] > upper_bound)]
print(f"Outliers in Client Spend:\n {outliers[['client_spend']]}")

# Boxplot for Client Spend to visualize outliers
sns.boxplot(x=data['client_spend'])
plt.title('Boxplot of Client Spend')
plt.show()

```

```python
# Convert specific columns to numeric, forcing errors to NaN
data['client_spend'] = pd.to_numeric(data['client_spend'], errors='coerce')
data['total_engagement'] = pd.to_numeric(data['total_engagement'], errors='coerce')
data['content_views'] = pd.to_numeric(data['content_views'], errors='coerce')

# Check for any remaining non-numeric values
print(data[['client_spend', 'total_engagement', 'content_views']].dtypes)

```

```python
# Check for missing values
print(data[['client_spend', 'total_engagement', 'content_views']].isnull().sum())

# Fill missing values with mean (or other strategy)
data['client_spend'].fillna(data['client_spend'].mean(), inplace=True)
data['total_engagement'].fillna(data['total_engagement'].mean(), inplace=True)
data['content_views'].fillna(data['content_views'].mean(), inplace=True)

# Confirm no missing values remain
print(data[['client_spend', 'total_engagement', 'content_views']].isnull().sum())
```

```python
import numpy as np

# Log Transformation for Skewed Variables
data['log_client_spend'] = np.log1p(data['client_spend'])  # log1p to handle zero values
data['log_total_engagement'] = np.log1p(data['total_engagement'])
data['log_content_views'] = np.log1p(data['content_views'])

# Visualizing the transformed variable - log_client_spend
sns.histplot(data['log_client_spend'], kde=True)
plt.title('Log-Transformed Distribution of Client Spend')
plt.show()
```

```python
# Square Root Transformation for Skewed Variables
data['sqrt_client_spend'] = np.sqrt(data['client_spend'])
data['sqrt_total_engagement'] = np.sqrt(data['total_engagement'])
data['sqrt_content_views'] = np.sqrt(data['content_views'])

# Visualizing the transformed variable - sqrt_total_engagement
sns.histplot(data['sqrt_total_engagement'], kde=True)
plt.title('Square Root Transformed Distribution of Total Engagement')
plt.show()
```

```python
# Creating a new feature: Engagement Rate (Total Engagement per Content View)
data['engagement_rate'] = data['total_engagement'] / data['content_views']

# Creating a new feature: Return on Investment (ROI)
data['roi'] = data['total_media_value'] / data['client_spend']

# Visualizing the new feature - Engagement Rate
sns.histplot(data['engagement_rate'], kde=True)
plt.title('Distribution of Engagement Rate')
plt.show()
```

```python
# Display summary statistics for the new features
print(data[['engagement_rate', 'roi', 'log_client_spend', 'sqrt_client_spend']].describe())
```
